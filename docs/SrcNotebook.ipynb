## Purpose 
- Data: Allows us to determine where our data will come from (API, CSV, Database, Etc).
- Model: Purpose is to define the class attributes for the type of data we are looking to handle.
- Service: Defines how we will able to work with the data we've obtained (logic).
- UI: How do i represent the results? (Console now, REACT later).
- Main: Wires everything together and starts the application.
