- Change approach from scraping to ingestion
    * A standard scraper requests a web page, pulls the data, and stops. This is fine for historical, completed auction results.
    * Scraping raw html from sites like copart or IAAI is difficult becaude they use heavy Javascript, bot-detection tools, and frequent UI updates.

    * Live auctions change by the second. An ingestion pipeline handles continuous, real-time data streams (like current bid prices, countdown timers, and vehicle status updates).
    * An ingestion pipeline prioritizes structured sources. It hooks directly into an official API or 3rd party vehicle data provider. Scraping is more of a last resort backup for fields the API doesn't provide.

    *Ingestion includes a crititical step called transformation (auto parses raw data, sanitizes it, and pings a secondary API to decode the vin #).
